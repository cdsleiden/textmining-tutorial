# 3. Word frequencies

### Exercise 3.1.

What are the 25 most frequent words in *Brave New World*, if you disregard all the tokens that are on the `nltk` list of English stopwords?

```python
import os
from os.path import join
from text_mining import *
from nltk import word_tokenize
from nltk.corpus import stopwords
from collections import Counter

# This code 'moves down' in the folder system
# because the folder 'Corpus' is on the same level as 'Solutions'
path_parent = os.path.dirname(os.getcwd())
path = join( path_parent, 'Corpus' , 'PrideandPrejudice.txt')

fh = open( path , encoding = 'utf-8' )
full_text = fh.read()

stopwords = stopwords.words('english')


words = word_tokenize(full_text.lower())
words = remove_punctuation_and_stopwords(words)


words_freq = Counter(words)

for word,count in words_freq.most_common(25):
    print(f"{word} => {count}")
    
```

### Exercise 3.2

In *Brave New World*, what is the most frequent word consisting of six characters? Show the words only if they have a frequency of more than 30. 

```python
import os
from text_mining import *
from nltk import word_tokenize
from nltk.corpus import stopwords
from collections import Counter

path_parent = os.path.dirname(os.getcwd())
path = join( path_parent, 'Corpus' , 'BraveNewWorld.txt')


fh = open( path , encoding = 'utf-8' )
full_text = fh.read()

stopwords = stopwords.words('english')


words = word_tokenize(full_text.lower())
words = remove_punctuation_and_stopwords(words)


words_freq = Counter(words)

for word,count in words_freq.most_common():
    if len(word)==6 and count>30:
        print(f"{word} => {count}")
    
```
