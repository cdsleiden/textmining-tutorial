# 1. Acquiring texts

### Exercise 1.1


Download all the text files that are listed in the following dictionary. 

`
    gutenberg_files = {
    'https://www.gutenberg.org/files/98/98-0.txt' :
        'A Tale of Two Cities',
    'https://www.gutenberg.org/files/580/580-0.txt':
        'The Pickwick Papers'
   }
`

Save these files in a folder named 'Texts'. In Python, you can make new folders using the `os` package, as follows:

`
os.mkdir('Texts')
`



```python
import os
import requests
from text_mining import *
import re

out_folder = 'Texts'

gutenberg_files = {
    'https://www.gutenberg.org/files/98/98-0.txt' :
        'A Tale of Two Cities',
    'https://www.gutenberg.org/files/580/580-0.txt':
        'The Pickwick Papers'
   }

if not os.path.exists(out_folder): 
    os.mkdir(out_folder)
    

for url in gutenberg_files:
    print("Downloading " + gutenberg_files[url] + " ...")
    response = requests.get(url)
    title = re.sub( r'\s+' , '_' ,  gutenberg_files[url].lower() )
    title = f'{title}.txt'

    if response:
        response.encoding = 'utf-8'
        full_text = remove_pg_boilerplate(response.text)           
        out = open( os.path.join(out_folder,title) , 'w' , encoding = 'utf-8')
        out.write( full_text.strip() )
        out.close()

print('\nDone!')    
```
