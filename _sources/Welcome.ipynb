## Welcome

This tutorial explains a range of fundamental concepts and methods that can be used to extract meaningful information from collections of unstructured texts. It covers basic preparatory operations, such as tokenisation, lemmatisation and part of speech tagging. The course also discusses more advanced analytic approaches in the field of natural language processing, including sentiment analysis named entity recognition and text classification. The tutorial assumes a basic proficiency in Python. 

You can download all the notebooks written for this tutorial at the following address:

https://github.com/cdsleiden/textmining-tutorial/zipball/master/

When you download and unzip this folder, make sure that you put the unzipped folder (text-and-data-mining-master by default) in a location that you can access from the Jupyter Notebook environment.
